```python
import sys, os
sys.path.insert(0, '../src')
from data_loader import fetch_dark_stores, generate_demand_grid
from opportunity_model import (compute_nearest_store_distance, flag_covered_zones,
                               compute_dsoi, get_top_expansion_zones, estimate_order_demand)
from visualizer import create_opportunity_map, plot_coverage_analysis, plot_opportunity_scatter
```

```python
stores = fetch_dark_stores()
zones  = generate_demand_grid()
zones  = compute_nearest_store_distance(zones, stores)
zones  = flag_covered_zones(zones)
zones  = compute_dsoi(zones)
top10  = get_top_expansion_zones(zones, n=10)
demand = estimate_order_demand(top10)
top10  = top10.merge(demand[['zone_id','est_daily_orders']], on='zone_id', how='left')
print('✅ Pipeline complete')
```

## 4.1 Interactive Opportunity Map

```python
os.makedirs('../outputs', exist_ok=True)
m = create_opportunity_map(zones, stores, top10, output_path='../outputs/dark_store_opportunity_map.html')
print('Open outputs/dark_store_opportunity_map.html in your browser.')
```

## 4.2 Coverage Analysis Chart

```python
plot_coverage_analysis(zones, stores, output_path='../outputs/coverage_analysis.png')
```

## 4.3 Opportunity Scatter

```python
plot_opportunity_scatter(zones, top10, output_path='../outputs/opportunity_scatter.png')
```

## 4.4 Save Final Outputs

```python
top10.drop(columns='geometry').to_csv('../outputs/top10_expansion_zones.csv', index=False)
demand.to_csv('../outputs/order_demand_estimates.csv', index=False)
print('\n✅ All outputs saved:')
print('  → dark_store_opportunity_map.html')
print('  → coverage_analysis.png')
print('  → opportunity_scatter.png')
print('  → top10_expansion_zones.csv')
print('  → order_demand_estimates.csv')
```
