Loads existing dark store locations (Blinkit/Zepto/Instamart) and generates demand zones for Delhi-NCR.

```python
import sys, os
sys.path.insert(0, '../src')
from data_loader import fetch_dark_stores, generate_demand_grid
import matplotlib.pyplot as plt
```

## 1.1 Load Existing Dark Stores

```python
stores = fetch_dark_stores()
print(f'Total stores: {len(stores)}')
print(stores['brand'].value_counts())
stores.head()
```

```python
fig, ax = plt.subplots(figsize=(9, 9))
colors = {'Blinkit':'gold','Zepto':'purple','Instamart':'orange','Unknown':'gray'}
for brand, grp in stores.groupby('brand'):
    grp.plot(ax=ax, color=colors.get(brand,'gray'), markersize=14, label=brand, alpha=0.85)
ax.set_title('Existing Dark Stores — Delhi-NCR', fontsize=13, fontweight='bold')
ax.legend(); plt.tight_layout(); plt.show()
```

## 1.2 Generate Demand Zones

```python
zones = generate_demand_grid(n_zones=150)
print(f'Zones generated: {len(zones)}')
zones['zone_type'].value_counts()
```

```python
fig, ax = plt.subplots(figsize=(9, 9))
zones.plot(column='population_density', ax=ax, cmap='YlOrRd', legend=True,
           legend_kwds={'label':'Pop. Density (persons/km²)'})
stores.plot(ax=ax, color='blue', markersize=10, zorder=5, label='Dark Stores')
ax.set_title('Demand Zones + Existing Dark Stores', fontsize=13, fontweight='bold')
ax.legend(); plt.tight_layout(); plt.show()
```

```python
os.makedirs('../data/raw', exist_ok=True)
stores.to_file('../data/raw/dark_stores.geojson', driver='GeoJSON')
zones.to_file('../data/raw/demand_zones.geojson', driver='GeoJSON')
print('✅ Raw data saved.')
```
